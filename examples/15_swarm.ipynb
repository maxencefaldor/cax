# Swarm 1-D tag consensus

A line of cells votes for local tags (idle / lit / run / match). Isolated workers cannot see across a chunk cut; chained workers can. Hand-coded rule, not a trained NCA.

```python
import jax.numpy as jnp
import numpy as np

from cax.cs.swarm import Swarm

RUN, LIT = 2.0, 1.0
worker = 4
tags = np.zeros((8, 1), dtype=np.float32)
tags[worker - 2, 0] = RUN
tags[worker - 1, 0] = LIT
tags[worker, 0] = RUN
iso = Swarm(worker_size=worker, isolated=True)
chained = Swarm(worker_size=worker, isolated=False)
iso_out = np.asarray(iso(jnp.asarray(tags), num_steps=1))
ch_out = np.asarray(chained(jnp.asarray(tags), num_steps=1))
assert np.isclose(iso_out[worker - 1, 0], LIT)
assert np.isclose(ch_out[worker - 1, 0], RUN)
print("isolated keeps the cut, chained crosses it")
```
