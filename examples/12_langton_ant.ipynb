## Import

```python
import jax
import jax.numpy as jnp
import mediapy
from flax import nnx

from cax.cs.langton_ant import LangtonAnt, LangtonAntState
```

## Configuration

```python
seed = 0

num_steps = 11_000
spatial_dims = (64, 64)
rule_string = "RL"  # Classic Langton's Ant

rngs = nnx.Rngs(seed)
```

## Instantiate system

```python
turns = LangtonAnt.turns_from_rule_string(rule_string)
turns
```

```python
cs = LangtonAnt(turns=turns, rngs=rngs)
```

## Sample initial state

```python
def sample_state():
	"""Sample a state with the ant at the center of an empty grid."""
	grid = jnp.zeros((*spatial_dims, 1))
	position = jnp.array([spatial_dims[0] // 2, spatial_dims[1] // 2], dtype=jnp.int32)
	direction = jnp.array(0, dtype=jnp.int32)  # North
	return LangtonAntState(grid=grid, position=position, direction=direction)
```

## Run

```python
state_init = sample_state()
state_final, states = cs(state_init, num_steps=num_steps, trajectory=True)
```

## Visualize

```python
states = jax.tree.map(
	lambda state_init, states: jnp.concatenate([state_init[None], states]),
	state_init,
	states,
)
frames = nnx.vmap(
	lambda cs, state: cs.render(state),
	in_axes=(None, 0),
)(cs, states)

mediapy.show_video(frames, width=256, height=256)
```

## Langton's Ant Family

You can experiment with other [generalized Langton's Ants](https://en.wikipedia.org/wiki/Langton%27s_ant#Extension_to_multiple_colors) by changing the rule string.

### LLRR

```python
turns = LangtonAnt.turns_from_rule_string("LLRR")
cs = LangtonAnt(turns=turns, rngs=rngs)
```

### LRRRRRLLR

```python
turns = LangtonAnt.turns_from_rule_string("LRRRRRLLR")
cs = LangtonAnt(turns=turns, rngs=rngs)
```

### Run

```python
state_init = sample_state()
state_final, states = cs(state_init, num_steps=num_steps, trajectory=True)
```

### Visualize

```python
states = jax.tree.map(
	lambda state_init, states: jnp.concatenate([state_init[None], states]),
	state_init,
	states,
)
frames = nnx.vmap(
	lambda cs, state: cs.render(state),
	in_axes=(None, 0),
)(cs, states)

mediapy.show_video(frames, width=256, height=256)
```
