# MQRI plate

A discrete automaton with a frozen payload strip. Rollout evolves the free field; **decode reads the strip** (payload rebuild). Identity hashes synthetic I/Q/U of the plate and detects tamper. Toy public rule — not a trained NCA.

```python
import jax.numpy as jnp
import numpy as np

from cax.cs.mqri import Mqri, decode_payload, encode_payload, reflection_id, strip_rows_for

payload = b"SLID-PHI|unit-pack|demo-2026-08-17"
rows = strip_rows_for(len(payload), 32)
seed = encode_payload(payload, height=32, width=32)
mqri = Mqri(height=32, width=32, strip_rows=rows)
final = np.asarray(mqri(jnp.asarray(seed), num_steps=12))
got = decode_payload(final, len(payload), strip_rows=rows)
assert got == payload
ident = reflection_id(final)
assert reflection_id(final) == ident
print("decode", got)
print("identity", ident)
```
